```python
import os
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE

# 1. Завантажуємо дані
data_path = '../data/raw/cs-training.csv' 
if not os.path.exists(data_path):
    data_path = 'data/raw/cs-training.csv'

df = pd.read_csv(data_path).iloc[:5000].drop(columns=['Unnamed: 0'], errors='ignore')
df = df.fillna(df.median(numeric_only=True))

# 2. Виділяємо цільову змінну
target_col = 'SeriousDlqin2yrs'
y = df[target_col].values

# 3. Створюємо захищену ознаку (на базі віку 'age') і ОДРАЗУ додаємо її в таблицю ознак X
df['protected_attribute'] = (df['age'] > df['age'].median()).astype(int)

# Забираємо цільову колонку та залишаємо все інше (включно з protected_attribute)
X = df.drop(columns=[target_col])

# 4. Робимо спліт
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

print("Дані успішно підготовлені!")
```

```python
# --- Вплив SMOTE на захищену ознаку ---
print("Розподіл до SMOTE:\n", X_train['protected_attribute'].value_counts(normalize=True))

smote = SMOTE(k_neighbors=5, random_state=42)
X_res, y_res = smote.fit_resample(X_train, y_train)

# Перетворюємо назад у DataFrame, щоб перевірити колонку після ресемплінгу
X_res_df = pd.DataFrame(X_res, columns=X_train.columns)
print("Розподіл після SMOTE:\n", X_res_df['protected_attribute'].value_counts(normalize=True))
```
